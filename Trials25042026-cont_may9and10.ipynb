```python
def fact(n):
    if(n==0):
        return 1

    return n * fact(n-1)

fact(5)
```

```python
def fact(n):
    if not isinstance(n,int):
        raise ValueError("Please provide integer")
    if(n==0):
        return 1
    return n * fact(n-1)

fact(1.8)
```
