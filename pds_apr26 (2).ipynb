```python
def fact(n):
    if not isinstance(n,int):
        raise ValueError("Boss. Please supply an Integer")
    
    if n < 0:
        raise ValueError("input should be positive integer")
    
    if n == 0:
        return 1
    
    return n * fact(n-1)
```
