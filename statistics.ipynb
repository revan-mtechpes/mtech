```python
#GMAT score follows normal distributions, mean = 680, var = 1600, standard deviation= 40 
```

```python
#1. What is the probability we will score more than 740?
#2. What is the probability we will score less than 680?
#3. What is the probability we will score between 720 to 760?
```

```python
#1. What is the probability we will score more than 740?
stats.norm.sf(740,680,40) #(Val,mean,sd)
```

```python
#2. What is the probability we will score less than 680?
stats.norm.cdf(680,680,40) #(Val,mean,sd) 
```

```python
#3. What is the probability we will score between 720 to 760?
print(stats.norm.cdf(760,680,40) - stats.norm.cdf(720,680,40)) 
print(stats.norm.sf(720,680,40) - stats.norm.sf(760,680,40))
print(1-stats.norm.cdf(720,680,40) - stats.norm.sf(760,680,40))
```
